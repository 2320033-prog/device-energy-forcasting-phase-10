```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score


# Step 1: Read data from the local file uploaded to Colab

file_path = 'household_power_consumption_trimmed.csv'
df = pd.read_csv(file_path, sep=';', low_memory=False)

print("Dataset Loaded Successfully!")
print(f"Dataset Shape: {df.shape}")
display(df.head())


# Step 2: Data cleaning and handling missing values

df.replace('?', np.nan, inplace=True)

numeric_cols = ['Global_active_power', 'Global_reactive_power', 'Voltage',
                'Global_intensity', 'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3']

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# ملء القيم المفقودة بالاستيفاء الخطي
df[numeric_cols] = df[numeric_cols].interpolate(method='linear')
df.dropna(subset=['Global_active_power'], inplace=True)

# دمج التاريخ والوقت واستخراج الميزات الزمنية
df['Datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], format='%d/%m/%Y %H:%M:%S')
df.set_index('Datetime', inplace=True)

df['Hour'] = df.index.hour
df['DayOfWeek'] = df.index.dayofweek
df['Lag_1'] = df['Global_active_power'].shift(1)
df.dropna(inplace=True)

print("Data Cleaning & Feature Engineering Complete!")
print(f"Cleaned Rows: {len(df)}")


# Step 3: Training the Linear Regression model

X = df[['Voltage', 'Global_intensity', 'Sub_metering_1', 'Sub_metering_2', 'Sub_metering_3', 'Hour', 'Lag_1']]
y = df['Global_active_power']

train_size = int(len(df) * 0.8)
X_train, X_test = X.iloc[:train_size], X.iloc[train_size:]
y_train, y_test = y.iloc[:train_size], y.iloc[train_size:]

model = LinearRegression()
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\n--- Baseline Linear Regression Results ---")
print(f"MAE: {mae:.4f} kW")
print(f"RMSE: {rmse:.4f} kW")
print(f"R² Score: {r2:.4f}")


# Step 4: Plotting the results (actual vs. predicted values)

plt.figure(figsize=(14, 5))
plt.plot(y_test.index[:150], y_test.values[:150], label='Actual Power', color='#1f77b4', linewidth=1.5)
plt.plot(y_test.index[:150], y_pred[:150], label='Predicted Power', color='#ff7f0e', linestyle='--', linewidth=1.5)

plt.title('Baseline Linear Regression: Actual vs Predicted Energy Forecasting', fontsize=14, fontweight='bold')
plt.xlabel('Date & Time')
plt.ylabel('Active Power (kW)')
plt.legend()
plt.grid(True, linestyle=':', alpha=0.6)
plt.tight_layout()
plt.show()
```
